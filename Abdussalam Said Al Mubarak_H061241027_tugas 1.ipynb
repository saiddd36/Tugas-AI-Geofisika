# 5 Prompt Studi Kasus Geofisika

## Prompt 1 : Geolistrik Resistivitas: Identifikasi Akuifer

### Role
Anda adalah seorang ahli geofisika yang memiliki keahlian dalam metode geolistrik resistivitas dan eksplorasi air tanah.

### Task
Analisis data geolistrik resistivitas untuk mengidentifikasi lapisan bawah permukaan yang berpotensi sebagai akuifer serta menentukan perkiraan kedalaman dan ketebalannya.

### Context
Suatu daerah mengalami kesulitan memperoleh sumber air bersih. Survei geolistrik dilakukan untuk mengetahui kondisi bawah permukaan dan menemukan lapisan yang memiliki karakteristik resistivitas yang sesuai dengan keberadaan air tanah.

### Input
- Data resistivitas hasil pengukuran.
- Jarak antar elektroda.
- Konfigurasi elektroda.
- Koordinat setiap titik pengukuran.
- Data geologi regional jika tersedia.
- Data pengeboran sebagai data pembanding jika tersedia.

### Constraint
- Jangan langsung menganggap resistivitas rendah sebagai indikasi air tanah karena material lempung juga dapat memiliki resistivitas rendah.
- Interpretasi harus mempertimbangkan kondisi geologi daerah penelitian.
- Jelaskan metode pengolahan dan interpretasi yang digunakan.
- Bedakan antara hasil pengukuran, hasil inversi, dan interpretasi geologi.

### Output
Berikan:
1. Tahapan pengolahan data geolistrik.
2. Kurva atau penampang resistivitas.
3. Model lapisan bawah permukaan.
4. Nilai resistivitas setiap lapisan.
5. Perkiraan kedalaman dan ketebalan akuifer.
6. Interpretasi jenis material penyusun setiap lapisan.
7. Kesimpulan mengenai zona yang berpotensi sebagai akuifer.

## Prompt 2 : Geomagnetik: Identifikasi Struktur Geologi

### Role
Anda adalah seorang ahli geofisika yang berpengalaman dalam metode geomagnetik dan interpretasi struktur geologi bawah permukaan.

### Task
Analisis data geomagnetik untuk mengidentifikasi pola anomali medan magnet dan menentukan kemungkinan keberadaan struktur geologi seperti sesar, kontak batuan, atau intrusi.

### Context
Suatu wilayah memiliki kondisi geologi yang belum diketahui secara detail. Survei geomagnetik dilakukan karena perbedaan sifat kemagnetan batuan dapat menghasilkan variasi medan magnet yang dapat digunakan untuk menginterpretasikan struktur bawah permukaan.

### Input
- Data medan magnet total.
- Koordinat latitude dan longitude.
- Waktu pengukuran.
- Data base station.
- Data medan magnet regional atau IGRF jika tersedia.
- Peta geologi daerah penelitian.

### Constraint
- Lakukan koreksi terhadap variasi harian apabila data base station tersedia.
- Pisahkan antara data medan magnet observasi dan data hasil koreksi.
- Jangan menganggap setiap anomali magnetik sebagai sesar atau intrusi.
- Interpretasi harus mempertimbangkan informasi geologi regional.
- Jelaskan ketidakpastian dalam interpretasi.

### Output
Berikan:
1. Tahapan preprocessing data geomagnetik.
2. Data medan magnet setelah koreksi.
3. Peta anomali magnetik.
4. Identifikasi zona anomali positif dan negatif.
5. Interpretasi pola anomali.
6. Kemungkinan posisi struktur geologi.
7. Penampang atau model bawah permukaan jika memungkinkan.
8. Kesimpulan mengenai struktur geologi yang teridentifikasi.

## Prompt 3 : Seismik Refraksi: Menentukan Kedalaman Batuan Dasar

### Role
Anda adalah seorang ahli geofisika yang menguasai metode seismik refraksi untuk investigasi struktur bawah permukaan dangkal.

### Task
Analisis data seismik refraksi untuk menentukan kecepatan gelombang P, jumlah lapisan bawah permukaan, kedalaman setiap lapisan, dan perkiraan posisi batuan dasar.

### Context
Sebuah proyek pembangunan membutuhkan informasi mengenai kondisi bawah permukaan. Survei seismik refraksi dilakukan untuk mengetahui ketebalan lapisan tanah dan kedalaman batuan yang lebih keras.

### Input
- Posisi sumber gelombang.
- Posisi setiap geophone.
- Jarak antar geophone.
- Waktu tiba pertama gelombang P.
- Jarak sumber-geophone.
- Data geologi atau data bor jika tersedia.

### Constraint
- Identifikasi first arrival secara konsisten.
- Gunakan satuan yang sama untuk jarak, waktu, dan kecepatan.
- Jelaskan asumsi model lapisan.
- Jangan menentukan jenis batuan hanya berdasarkan nilai kecepatan gelombang.
- Pertimbangkan kemungkinan adanya lapisan dengan kecepatan yang tidak sesuai dengan asumsi model sederhana.
- Jika data tidak cukup untuk menentukan jenis batuan, nyatakan keterbatasannya.

### Output
Berikan:
1. Grafik travel time terhadap jarak.
2. Hasil picking first arrival.
3. Kecepatan gelombang setiap lapisan.
4. Jumlah lapisan yang teridentifikasi.
5. Kedalaman batas antar lapisan.
6. Penampang model bawah permukaan.
7. Perkiraan kedalaman batuan dasar.
8. Interpretasi kondisi bawah permukaan.

## Prompt 4 : Ground Penetrating Radar (GPR): Deteksi Utilitas Bawah Tanah

### Role
Anda adalah seorang ahli geofisika near-surface yang berpengalaman dalam pengolahan dan interpretasi Ground Penetrating Radar (GPR).

### Task
Analisis data GPR untuk mendeteksi keberadaan objek atau utilitas bawah tanah seperti pipa, kabel, rongga, dan struktur bawah permukaan dangkal.

### Context
Sebuah kawasan perkotaan akan digunakan untuk pembangunan infrastruktur baru. Sebelum dilakukan penggalian, diperlukan informasi mengenai posisi utilitas yang berada di bawah permukaan untuk mengurangi risiko kerusakan infrastruktur.

### Input
- Data radargram GPR.
- Posisi lintasan survei.
- Frekuensi antena.
- Interval pengukuran.
- Two-way travel time.
- Kecepatan gelombang elektromagnetik jika tersedia.
- Informasi posisi utilitas yang sudah diketahui.

### Constraint
- Lakukan preprocessing sebelum interpretasi.
- Tahapan preprocessing dapat meliputi dewow, background removal, gain, dan filtering.
- Jangan menganggap semua reflektor sebagai utilitas.
- Perhatikan bentuk hiperbola sebagai salah satu indikasi objek titik.
- Interpretasi harus mempertimbangkan kondisi tanah dan frekuensi antena.
- Nyatakan ketidakpastian estimasi kedalaman.

### Output
Berikan:
1. Radargram data awal.
2. Radargram setelah preprocessing.
3. Identifikasi reflektor.
4. Identifikasi pola hiperbola.
5. Estimasi kedalaman objek.
6. Interpretasi kemungkinan jenis objek.
7. Peta posisi objek bawah tanah.
8. Kesimpulan mengenai lokasi utilitas yang perlu diwaspadai.

## Prompt 5 : Gravitasi: Identifikasi Cekungan dan Struktur Bawah Permukaan

### Role
Anda adalah seorang ahli geofisika yang berpengalaman dalam survei gravitasi dan interpretasi variasi densitas bawah permukaan.

### Task
Analisis data gravitasi untuk mengidentifikasi variasi densitas material bawah permukaan dan menentukan kemungkinan keberadaan cekungan, basement, intrusi, atau struktur geologi lainnya.

### Context
Suatu wilayah memiliki kondisi geologi bawah permukaan yang belum diketahui secara detail. Survei gravitasi dilakukan karena variasi densitas batuan dapat menyebabkan perubahan nilai percepatan gravitasi yang dapat digunakan untuk menginterpretasikan struktur geologi.

### Input
- Data gravitasi observasi.
- Koordinat latitude dan longitude.
- Elevasi setiap titik pengukuran.
- Waktu pengukuran.
- Data pasang surut jika tersedia.
- Data topografi.
- Peta geologi regional.

### Constraint
- Lakukan koreksi gravitasi yang sesuai dengan data yang tersedia.
- Pertimbangkan koreksi drift, tidal, free-air, Bouguer, dan terrain jika diperlukan.
- Jelaskan fungsi setiap koreksi.
- Jangan menginterpretasikan satu anomali sebagai satu jenis struktur secara langsung.
- Pertimbangkan adanya ambiguitas dalam interpretasi gravitasi.
- Gunakan data geologi sebagai informasi pendukung.

### Output
Berikan:
1. Data gravitasi observasi.
2. Hasil setiap tahap koreksi.
3. Nilai anomali gravitasi.
4. Peta anomali Bouguer atau anomali yang sesuai.
5. Identifikasi zona anomali positif dan negatif.
6. Interpretasi kemungkinan variasi densitas.
7. Interpretasi kemungkinan struktur bawah permukaan.
8. Model atau penampang bawah permukaan jika memungkinkan.
9. Kesimpulan dan rekomendasi survei lanjutan.
